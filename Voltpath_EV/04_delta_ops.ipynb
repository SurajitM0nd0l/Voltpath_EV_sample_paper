```python
spark.sql("""
CREATE OR REPLACE TABLE IDENTIFIER("{catalog}.ops.time_travel_check") AS
SELECT
    a.charger_id,
    a.max_kw      AS max_kw_before_batch2,
    b.max_kw      AS max_kw_current,
    a.is_active   AS is_active_before_batch2,
    b.is_active   AS is_active_current
FROM silver.chargers VERSION AS OF 1 a
JOIN silver.chargers b
  ON a.charger_id = b.charger_id
WHERE a.charger_id IN ('CH004', 'CH011', 'CH027')
""")
```
